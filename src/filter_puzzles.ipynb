This is a helper notebook that extracts some pre-defined sudokus for use.
The original dataset can be retrieved directly from [here](https://www.kaggle.com/datasets/radcliffe/3-million-sudoku-puzzles-with-ratings).

I gather 30 sudokus from each sub-difficulty. With 8.5 being the highest and the step being at 0.1, that means that theoretically 2550 sudokus will be extracted. Practically the number will be a little bit less as some performance brackets are not fully saturated.

```python
# Imports
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
file  = pd.read_csv("sudoku-3m.csv")
file.head()
```

```python
#Sort by difficulty for ease, although not needed
file = file.sort_values(by=["difficulty"])
```

Now i extract the data with a simple for loop, each for a different difficulty bracket

```python
target_df = pd.DataFrame()
last_valid = 0.0
for i in np.round(np.arange(0.0, 8.6, 0.1), 1):
    subset_df = file[file.difficulty == i]
    if len(subset_df) > 0:
        target_df = pd.concat([target_df, subset_df.iloc[:30]], ignore_index=True)
        last_valid = i
    else:
        subset_df = file[file.difficulty == last_valid]
        target_df = pd.concat([target_df, subset_df.iloc[:30]], ignore_index=True)
```

```python
# Quick check that the final puzzle number checks out
len(target_df)
```

```python
# Create the folder (and any parent folders) if they don't already exist
output_dir = Path("../data")
output_dir.mkdir(parents=True, exist_ok=True)

target_df[["puzzle"]].to_csv("../data/puzzles.txt", sep=" ", header=False, index=False)
target_df[["solution"]].to_csv("../data/solutions.txt", sep=" ", header=False, index=False)
```
